````python
import pandas as pd
import numpy as np

# Create sample reactor data
np.random.seed(42)
n = 100
df = pd.DataFrame({
    'reactor': np.random.choice(['R1', 'R2', 'R3'], n),
    'temperature': np.random.uniform(300, 500, n),
    'pressure': np.random.uniform(1, 10, n),
    'conversion': np.random.uniform(0.3, 0.95, n),
    'shift': np.random.choice(['Day', 'Night'], n)
})
df.loc[np.random.choice(n, 5), 'conversion'] = np.nan  # Add some missing values

# QUESTIONS - Answer using pandas operations:
# 1. How many rows and columns?
# 2. How many missing values in 'conversion'?
# 3. What is the mean temperature for reactor R2?
# 4. How many experiments were run on the Night shift?
# 5. What is the maximum conversion?

# Your code here:
````

````python
# Ref codes:
rows, columns = df.shape # recal what .shap returns
missing_conversion = df['conversion'].isna().sum() # recall what .isna() returns
mean_temperature_r2 = df.loc[df['reactor'] == 'R2', 'temperature'].mean()
night_count = (df['shift'] == 'Night').sum()
max_conversion = df['conversion'].max()

print(f"1. Rows: {rows}; columns: {columns}")
print(f"2. Missing conversion values: {missing_conversion}")
print(f"3. Mean temperature for R2: {mean_temperature_r2:.6f}")
print(f"4. Night-shift experiments: {night_count}")
print(f"5. Maximum conversion: {max_conversion:.9f}")
print("Valid conversion measurements:", df['conversion'].count())
assert df.shape == (100, 5)
assert df['conversion'].count() + missing_conversion == len(df)
````
